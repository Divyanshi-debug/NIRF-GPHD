```python
import pandas as pd
import numpy as np
from scipy.optimize import curve_fit

df23 = pd.read_csv('gphd_2023_working.csv')  # change filename
df23['nphd_ft'] = df23['total_ft'] / 3
df23['f'] = df23['GPHD'] / 20

x = df23['nphd_ft'].values
y = df23['f'].values

# fix p=0.62, find X
def power_model_fixed_p(x, X):
    return np.minimum((x / X) ** 0.62, 1)

params, _ = curve_fit(power_model_fixed_p, x, y, p0=[300])

y_pred = power_model_fixed_p(x, params[0])
ss_res = np.sum((y - y_pred) ** 2)
ss_tot = np.sum((y - y.mean()) ** 2)
r2 = 1 - (ss_res / ss_tot)

print(f"2023 → X = {params[0]:.2f}, R² = {r2:.4f}")
```
